```python
def camel_to_snake(name):
    # CamelCase → snake_case
    s = re.sub('(.)([A-Z][a-z]+)', r'\1_\2', name)
    return re.sub('([a-z0-9])([A-Z])', r'\1_\2', s).lower()

df.columns = [camel_to_snake(col) for col in df.columns]
```
